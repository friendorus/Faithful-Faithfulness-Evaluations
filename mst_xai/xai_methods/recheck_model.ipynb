```python
import torch
from pathlib import Path
from mst.models.dino import DinoV2ClassifierSlice

ckpt_path = Path("runs/Local/DinoV2ClassifierSlice_Final")
model = DinoV2ClassifierSlice.load_best_checkpoint(ckpt_path)

model.eval()
```

```python
import torch

def shape_hook(module, inp, out):
    if isinstance(out, torch.Tensor):
        print(f"{module.__class__.__name__}: {out.shape}")

hooks = []
for name, module in model.named_modules():
    hooks.append(module.register_forward_hook(shape_hook))

# run ONE forward pass
dummy = torch.randn(1, 3, 32, 224, 224)  # adapt if needed
with torch.no_grad():
    _ = model(dummy)

# remove hooks
for h in hooks:
    h.remove()
```
