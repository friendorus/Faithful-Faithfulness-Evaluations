```python
import torch
from pathlib import Path
import pandas as pd

saliency_dir = Path(
    "/home/jovyan/work/MST/results/DINOv3ViTB/saliency_results/random"
)

rows = []

for pt_file in saliency_dir.glob("class_*/pt/*_importance.pt"):

    sal = torch.load(pt_file)

    rows.append({
        "uid": pt_file.stem.replace("_importance", ""),
        "min": sal.min().item(),
        "max": sal.max().item(),
        "mean": sal.mean().item(),
        "std": sal.std().item(),
    })

df = pd.DataFrame(rows)

print(df.describe())
```

```python
import torch
import pandas as pd
from pathlib import Path

root = Path("/home/jovyan/work/MST/results/DINOv3ViTB/saliency_results")

rows = []

for method_dir in root.iterdir():

    if not method_dir.is_dir():
        continue

    files = list(method_dir.rglob("*_importance.pt"))

    print(f"{method_dir.name}: {len(files)} files")

    for f in files:

        sal = torch.load(f, map_location="cpu")

        rows.append({
            "method": method_dir.name,
            "min": sal.min().item(),
            "max": sal.max().item(),
            "mean": sal.mean().item(),
            "std": sal.std().item(),
        })

df = pd.DataFrame(rows)

summary = (
    df.groupby("method")[["min", "max", "mean", "std"]]
      .agg(["mean", "std", "min", "max"])
)

print(summary)
```
